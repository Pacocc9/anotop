## 0 · Preparación

No toques esta celda salvo para añadir librerías. **Nada de rutas absolutas** (`C:\Users\...`): el jurado ejecutará el notebook en otro ordenador.

```python
%%time
import time
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

INICIO = time.perf_counter()     # para medir el tiempo total (InnoSoft Reciclator)
URL_BASE = "https://innosoft-igualdad-sostenibilidad.vercel.app/data/csv/"


def leer(nombre_fichero):
    """Lee un CSV del datatón (local si existe, si no desde la web)."""
    for carpeta in [Path("datos/csv"), Path("../datos/csv")]:
        if (carpeta / nombre_fichero).exists():
            return pd.read_csv(carpeta / nombre_fichero, sep=";", decimal=",", thousands=".",
                               encoding="utf-8-sig", na_values=["."])
    return pd.read_csv(URL_BASE + nombre_fichero, sep=";", decimal=",", thousands=".",
                       encoding="utf-8-sig", na_values=["."])


def brecha(h, m):
    """Brecha en % con las dos definiciones."""
    return {"brecha_sobre_hombres_pct": (h - m) / h * 100, "brecha_hombres_sobre_mujeres_pct": (h / m - 1) * 100}


Path("figuras").mkdir(exist_ok=True)
Path("resultados").mkdir(exist_ok=True)
```

## 1 · Carga y limpieza

✏️ *Lee aquí cada fichero UNA vez. Documenta cada decisión de limpieza en un comentario: NaN, filas de totales, etiquetas, valores imposibles.*

```python
# Ejemplo (bórralo y pon los tuyos)
datos = leer("DS-005_salario_sexo_andalucia_2024.csv")
print(datos.shape)
print("Valores ausentes por columna:\n", datos.isna().sum()[lambda s: s > 0])
datos.head()
```

## 2 · Análisis

✏️ *Indicadores que responden a tu pregunta. Compara media y mediana. Si usas DS-013, recuerda que es sintético.*

```python
# Ejemplo
s = datos.set_index("sexo")
resultados = []
for col in ["salario_bruto_anual", "salario_eq_tiempo_completo"]:
    fila = {"indicador": col, "fuente": "DS-005 · IECA MCVL 2024 · Andalucía", "unidad": "€",
            "hombres": s.loc["Hombres", col], "mujeres": s.loc["Mujeres", col]}
    fila.update(brecha(fila["hombres"], fila["mujeres"]))
    resultados.append(fila)

tabla_resultados = pd.DataFrame(resultados)
tabla_resultados.round(2)
```

## 3 · Gráficos (3–4)

✏️ *Cada gráfico: título-conclusión, ejes con unidades, fuente. Se guardan en `figuras/`.*

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.barh(tabla_resultados["indicador"], tabla_resultados["brecha_sobre_hombres_pct"], color="#7a5195")
ax.set_xlabel("Brecha (H − M) / H, en %")
fig.suptitle("TÍTULO QUE ES UNA CONCLUSIÓN", x=0.01, ha="left", fontweight="bold")
fig.text(0.01, 0.01, "Fuente: …", fontsize=8, color="gray")
fig.tight_layout(rect=(0, 0.04, 1, 1))
fig.savefig("figuras/grafico_1.png", dpi=150)
plt.show()
```
