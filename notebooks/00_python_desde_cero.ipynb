## 2. Variables: ponerle nombre a un valor

Una **variable** es una etiqueta para guardar un valor y usarlo después. Se crea con `=`.

````python
salario_hombres = 21103     # salario bruto anual medio, Andalucía 2024 (€)
salario_mujeres = 17093

diferencia = salario_hombres - salario_mujeres
print(diferencia)
````

### Tipos de datos básicos

| Tipo | Ejemplo | Para qué |
|---|---|---|
| `int` (entero) | `2024` | contar, años |
| `float` (decimal) | `19.0` | salarios, tasas |

⚠️ En Python el decimal se escribe con **punto**: `19.5`, no `19,5`. (Los CSV del datatón usan coma; por eso al leerlos hay que avisar a pandas.)

## 3. Texto con números dentro: *f-strings*

Pon una `f` delante de las comillas y mete variables entre `{}`. Después de `:` decides el formato:
`:.1f` → un decimal · `:.1%` → porcentaje con un decimal · `:,.0f` → separador de miles.

````python
brecha = (salario_hombres - salario_mujeres) / salario_hombres
print(f"Los hombres cobran {salario_hombres:,.0f} € y las mujeres {salario_mujeres:,.0f} €")
print(f"La brecha es del {brecha:.1%}")
````

## 4. Listas y diccionarios

- Una **lista** guarda varios valores en orden: `[a, b, c]`.
- Un **diccionario** guarda pares *clave → valor*: `{"Hombres": 21103}`.

````python
grupos_edad = ["Menos de 30", "30-39", "40-49", "50-59", "60 y más"]
print(grupos_edad[0])        # el primero (¡se empieza a contar en 0!)
print(len(grupos_edad))      # cuántos hay

salario_por_sexo = {"Hombres": 21103, "Mujeres": 17093}
print(salario_por_sexo["Mujeres"])
````

## 5. Bucles (`for`) y condiciones (`if`)

Un **bucle** repite algo para cada elemento. Una **condición** decide qué hacer.
La **sangría** (los 4 espacios) indica qué está "dentro" del bucle o de la condición.

````python
tasas_temporalidad = {"Almería": 18.05, "Cádiz": 21.65, "Jaén": 25.80, "Málaga": 15.01}

for provincia, tasa in tasas_temporalidad.items():
    if tasa > 20:
        print(f"{provincia}: {tasa} % ⚠️ por encima del 20 %")
    else:
        print(f"{provincia}: {tasa} %")
````

## 6. Funciones: escribir una vez, usar muchas

Una **función** es un trozo de código con nombre que recibe datos y devuelve un resultado. Se define con `def`.

````python
def brecha_sobre_hombres(h, m):
    """(H - M) / H: qué parte del salario masculino 'falta' en el femenino."""
    return (h - m) / h


def brecha_hombres_sobre_mujeres(h, m):
    """H / M - 1: cuánto cobran de más los hombres respecto a las mujeres."""
    return h / m - 1


print(f"{brecha_sobre_hombres(21103, 17093):.1%}")
print(f"{brecha_hombres_sobre_mujeres(21103, 17093):.1%}")
````

👀 **Fíjate:** con los mismos números salen **19,0 %** y **23,5 %**. Las dos son correctas: miden cosas distintas. Por eso siempre hay que decir **qué fórmula usas**.

## 8. pandas en 5 minutos

La pieza central de pandas es el **DataFrame**: una tabla con **filas** (observaciones) y **columnas** (variables). Cada columna es una **Series**.

Vamos a crear una tabla pequeña a mano para practicar (datos inventados).

````python
tabla = pd.DataFrame({
    "nombre":   ["Ana", "Luis", "Marta", "Pedro", "Lucía", "Javi"],
    "sexo":     ["Mujer", "Hombre", "Mujer", "Hombre", "Mujer", "Hombre"],
    "edad":     [25, 31, 45, 52, 38, 29],
    "contrato": ["Temporal", "Indefinido", "Indefinido", "Indefinido", "Temporal", "Temporal"],
    "salario":  [18000, 26000, 31000, 90000, 24000, 21000],
})
tabla
````

````python
# Elegir una columna
tabla["salario"]
````

````python
# Filtrar filas: solo las mujeres
tabla[tabla["sexo"] == "Mujer"]
````

````python
# Crear una columna nueva a partir de otra (sin bucles: pandas lo hace de golpe)
tabla["salario_mensual"] = tabla["salario"] / 12
tabla
````

````python
# Agrupar: media y mediana del salario por sexo
tabla.groupby("sexo")["salario"].agg(["mean", "median", "count"])
````

### 🧠 Primera lección de estadística

Mira la media y la mediana de los hombres. **Pedro (90.000 €) tira de la media hacia arriba**; la mediana casi no se mueve.

- **Media**: suma ÷ número de personas. Sensible a valores extremos.
- **Mediana**: el valor del medio si ordenas a todo el mundo. Representa mejor a "la persona típica".

👉 En salarios, **compara siempre media y mediana**.

## 9. Un gráfico en 4 líneas

````python
medias = tabla.groupby("sexo")["salario"].median()

fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(medias.index, medias.values, color=["#3b6fb6", "#d9822b"])
ax.set_title("Salario mediano por sexo (datos inventados)")
ax.set_ylabel("€ brutos al año")
plt.show()
````

## 10. Cuando algo falla (y fallará)

Los errores son normales. Lee **la última línea** del mensaje rojo: suele decir qué pasa.

| Error | Qué significa | Solución típica |
|---|---|---|
| `NameError: name 'x' is not defined` | Usas una variable que no existe | ¿Has ejecutado la celda donde se crea? ¿Está bien escrita? |
| `KeyError: 'salario'` | Esa columna no existe | Mira `df.columns`: puede tener otro nombre, tildes o espacios |
| `TypeError: unsupported operand` | Operas con texto como si fuera número | Revisa `df.dtypes`: ¿la columna se leyó como texto? |
| `FileNotFoundError` | No encuentra el fichero | Revisa la ruta (y que sea relativa) |
| `IndentationError` | Sangría mal | Usa 4 espacios dentro de `for`, `if`, `def` |
